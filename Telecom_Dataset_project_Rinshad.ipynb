```python

plt.figure(figsize = (10,5))
sns.scatterplot(data = data, x = 'RAM', y = 'Price', color = 'blue', edgecolor = 'linen', alpha = 0.5)
plt.title("Scatterplot of RAM vs Price")
plt.xlabel('RAM')
plt.ylabel('Price')
plt.show()
```

```python

plt.figure(figsize = (10,5))
sns.scatterplot(data = data, x = 'Depth', y = 'Weight', color = 'red', edgecolor = 'linen', alpha = 0.5)
plt.title("Scatterplot of depth vs weight")
plt.xlabel('Depth')
plt.ylabel('Weight')
plt.show()
```

```python
plt.figure(figsize = (10,5))
sns.scatterplot(data = data, x = 'RAM', y = 'Weight', color = 'k', edgecolor = 'linen', alpha = 0.5)
plt.title("Scatterplot of RAM vs Weight")
plt.xlabel('RAM')
plt.ylabel('Weight')
plt.show()
```

```python
plt.figure(figsize = (10,5))
sns.countplot(data = data, x = 'Blue', edgecolor = 'linen', alpha = 0.7)
plt.title("Countplot of Blue")
plt.xlabel('Blue')
plt.ylabel('Count')
plt.show()
```

```python

plt.figure(figsize = (10,5))
sns.histplot(data = data, x = 'Price', color = 'orange', edgecolor = 'linen', alpha = 0.7, bins = 10)
plt.title("Histogram of price")
plt.xlabel('Price')
plt.ylabel('Count')
plt.show()
```
